```python
# ==============================================================================
# CELL 3: Kernel Audit & Formal Verification Suite for ATM (12 Modules)
# (Algebraic Theory of Modular Projection Sieving)
# ==============================================================================
import hashlib
import os
import re
import shutil
import subprocess
import sys
import zipfile
from datetime import datetime, timezone

PROJECT_DIR = "/content/ATM_Lean"
SRC_DIR = os.path.join(PROJECT_DIR, "ATM_Lean")
ZIP_OUTPUT_PATH = "/content/ATM_Lean4_Reproducibility_Suite.zip"
env_path = f"/root/.elan/bin:{os.environ.get('PATH', '')}"

os.makedirs(SRC_DIR, exist_ok=True)

# ------------------------------------------------------------------------------
# 1. WRITING THE 12 LEAN 4 FORMALIZATION MODULES (AUTONOMOUS / ZERO SORRY)
# ------------------------------------------------------------------------------

MODULES = {

# ============================================================
# MODULE 1: ModularClassification (N1)
# Thm 2.1 — Modular classification over Z/6Z
# ============================================================
"ModularClassification": """import Mathlib.Data.Nat.Basic
import Mathlib.Data.Int.Basic
import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Tactic

namespace ATM.Classification

def C1 (k : ℤ) : ℤ := 6 * k + 1
def C5 (k : ℤ) : ℤ := 6 * k - 1

theorem class_0_even (k N : ℕ) (h : N = 6 * k) : ∃ m, N = 2 * m :=
  ⟨3 * k, by omega⟩

theorem class_0_div_three (k N : ℕ) (h : N = 6 * k) : ∃ n, N = 3 * n :=
  ⟨2 * k, by omega⟩

theorem class_2_even (k N : ℕ) (h : N = 6 * k + 2) : ∃ m, N = 2 * m :=
  ⟨3 * k + 1, by omega⟩

theorem class_3_div_three (k N : ℕ) (h : N = 6 * k + 3) : ∃ n, N = 3 * n :=
  ⟨2 * k + 1, by omega⟩

theorem class_4_even (k N : ℕ) (h : N = 6 * k + 4) : ∃ m, N = 2 * m :=
  ⟨3 * k + 2, by omega⟩

/-- §2.1 (N1): every prime p > 3 lies in C1 or C5, i.e., p ≡ 1 or 5 (mod 6). -/
theorem primes_gt_three_in_channels (p : ℕ) (hp : Nat.Prime p) (hp3 : 3 < p) :
    p % 6 = 1 ∨ p % 6 = 5 := by
  have h_mod : p % 6 = 0 ∨ p % 6 = 1 ∨ p % 6 = 2 ∨ p % 6 = 3 ∨ p % 6 = 4 ∨ p % 6 = 5 := by omega
  rcases h_mod with h0 | h1 | h2 | h3 | h4 | h5
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6), by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exact Or.inl h1
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6) + 1, by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exfalso
    have h_div : 3 ∣ p := ⟨2 * (p / 6) + 1, by omega⟩
    have hp_eq_3 : 3 = p := (hp.eq_one_or_self_of_dvd 3 h_div).resolve_left (by decide)
    omega
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6) + 2, by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exact Or.inr h5

end ATM.Classification
""",

# ============================================================
# MODULE 2: PositionalCongruence (N1)
# Lemma 3.1 — Positional congruence
# ============================================================
"PositionalCongruence": """import Mathlib.Data.Int.Basic
import Mathlib.Algebra.GCDMonoid.Basic
import Mathlib.Tactic

namespace ATM.Positional

/-- Auxiliary lemma: Euclid's lemma via Bézout's identity. -/
lemma dvd_of_dvd_mul_6 (p X : ℤ)
    (h1 : ∃ c : ℤ, 6 * X = p * c)
    (hcoprime : ∃ a b : ℤ, a * p + b * 6 = 1) :
    ∃ d : ℤ, X = p * d := by
  rcases hcoprime with ⟨a, b, hab⟩
  rcases h1 with ⟨c, hc⟩
  use a * X + b * c
  calc X
    _ = X * 1 := by ring
    _ = X * (a * p + b * 6) := by rw [hab]
    _ = a * X * p + b * (6 * X) := by ring
    _ = a * X * p + b * (p * c) := by rw [hc]
    _ = p * (a * X + b * c) := by ring

/-- Lemma 3.1 (corrected sign): p | 6k + eps ↔ p | (k - eps*ep*kp). -/
theorem positional_congruence (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (hcop : IsCoprime (6 : ℤ) p) :
    (∃ c, 6 * k + eps = p * c) ↔
    (∃ d, k - ep * eps * kp = p * d) := by
  have h_alg : (6 * k + eps) - p * (ep * eps) = 6 * (k - ep * eps * kp) := by
    calc (6 * k + eps) - p * (ep * eps)
      _ = 6 * k + eps - (6 * kp + ep) * (ep * eps) := by rw [hp]
      _ = 6 * k + eps - 6 * kp * ep * eps - (ep * ep) * eps := by ring
      _ = 6 * k + eps - 6 * kp * ep * eps - 1 * eps := by rw [hep]
      _ = 6 * (k - ep * eps * kp) := by ring
  obtain ⟨a, b, hab⟩ := hcop
  have hbez : ∃ a b : ℤ, a * p + b * 6 = 1 := ⟨b, a, by linarith [hab]⟩
  constructor
  · intro h
    rcases h with ⟨c, hc⟩
    have h1 : 6 * (k - ep * eps * kp) = p * (c - ep * eps) := by
      calc 6 * (k - ep * eps * kp)
        _ = (6 * k + eps) - p * (ep * eps) := by rw [h_alg]
        _ = p * c - p * (ep * eps) := by rw [hc]
        _ = p * (c - ep * eps) := by ring
    exact dvd_of_dvd_mul_6 p (k - ep * eps * kp) ⟨c - ep * eps, h1⟩ hbez
  · intro h
    rcases h with ⟨d, hd⟩
    refine ⟨6 * d + ep * eps, ?_⟩
    calc 6 * k + eps
      _ = 6 * (k - ep * eps * kp) + p * (ep * eps) := by
        calc 6 * k + eps
          _ = (6 * k + eps) - p * (ep * eps) + p * (ep * eps) := by ring
          _ = 6 * (k - ep * eps * kp) + p * (ep * eps) := by rw [h_alg]
      _ = 6 * (p * d) + p * (ep * eps) := by rw [hd]
      _ = p * (6 * d + ep * eps) := by ring

/-- §3.1 corollary (N1): no annihilation ↔ no positional congruence.
    Contrapositive form of Lemma 3.1, used in §4.2 (ground state). -/
theorem no_annihilation_iff_no_congruence (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (hcop : IsCoprime (6 : ℤ) p) :
    ¬(∃ c, 6 * k + eps = p * c) ↔
    ¬(∃ d, k - ep * eps * kp = p * d) := by
  exact (not_iff_not).2 (positional_congruence k kp p ep eps hp hep hcop)

end ATM.Positional
""",

# ============================================================
# MODULE 3: KminPlus (N1)
# Thm 3.4 — Explicit formula for k_min^+
# ============================================================
"KminPlus": """import Mathlib.Data.Int.Basic
import Mathlib.Tactic.Ring

namespace ATM.KminPlus

def kminPlus (p kp ep : ℤ) : ℤ := p * kp + ep * kp

theorem kminPlus_spec (p kp ep : ℤ) (hp : p = 6 * kp + ep)
    (heps : ep * ep = 1) :
    6 * kminPlus p kp ep + 1 = p ^ 2 := by
  unfold kminPlus
  rw [hp]
  have h_sq : ep ^ 2 = 1 := by rw [sq]; exact heps
  have h : (6 * kp + ep) ^ 2 = 36 * kp ^ 2 + 12 * ep * kp + ep ^ 2 := by ring
  rw [h, h_sq]
  ring

end ATM.KminPlus
""",

# ============================================================
# MODULE 4: EntanglementThresholds (N1)
# Thm 3.7 — Four algebraic cases of kappa_min^-
# ============================================================
"EntanglementThresholds": """import Mathlib.Data.Int.Basic
import Mathlib.Tactic.Ring

namespace ATM.Entanglement

/-- Thm 3.7, case (p ∈ C5, q ∈ C1): kappa_min^- = p * kq + kp -/
theorem entanglement_minus_plus (p kp kq kmin : ℤ)
    (hp : p = 6 * kp - 1) (hkmin : kmin = p * kq + kp) :
    6 * kmin - 1 = p * (6 * kq + 1) := by
  rw [hkmin, hp]; ring

/-- Thm 3.7, case (p ∈ C1, q ∈ C5): kappa_min^- = p * kq - kp -/
theorem entanglement_plus_minus (p kp kq kmin : ℤ)
    (hp : p = 6 * kp + 1) (hkmin : kmin = p * kq - kp) :
    6 * kmin - 1 = p * (6 * kq - 1) := by
  rw [hkmin, hp]; ring

/-- Thm 3.7, case (p ∈ C5, q ∈ C5): kappa_min^+ = p * kq - kp -/
theorem entanglement_minus_minus (p kp kq kmin : ℤ)
    (hp : p = 6 * kp - 1) (hkmin : kmin = p * kq - kp) :
    6 * kmin + 1 = p * (6 * kq - 1) := by
  rw [hkmin, hp]; ring

/-- Thm 3.7, case (p ∈ C1, q ∈ C1): kappa_min^+ = p * kq + kp -/
theorem entanglement_plus_plus (p kp kq kmin : ℤ)
    (hp : p = 6 * kp + 1) (hkmin : kmin = p * kq + kp) :
    6 * kmin + 1 = p * (6 * kq + 1) := by
  rw [hkmin, hp]; ring

/-- Definition of entanglement jump Δk = kq - kp -/
def entanglementJump (kp kq : ℤ) : ℤ := kq - kp

/-- Twin-prime ground state condition: Δk = 0 iff kq = kp -/
theorem ground_state_iff_jump_zero (kp kq : ℤ) :
    entanglementJump kp kq = 0 ↔ kq = kp := by
  unfold entanglementJump; omega

end ATM.Entanglement
""",

# ============================================================
# MODULE 5: CopExistence (N1-P)
# Def 3.5 + Thm 3.6 — Positional coprime, existence via Dirichlet
# ============================================================
"CopExistence": """import Mathlib.Data.Int.Basic
import Mathlib.Data.Nat.Prime.Basic

namespace ATM.Cop

/-- Def 3.5: A prime q is a valid positional coprime of p = 6kp + ep
    if q lies in the opposite channel and its index satisfies the
    asymmetric condition. -/
def IsCop (p kp ep : ℤ) (q kq : ℤ) : Prop :=
  q = 6 * kq - ep ∧
  (if ep = -1 then kq ≥ kp else kq ≥ kp + 1)

/-- Def 3.5: The entanglement jump -/
def deltaK (kp kq : ℤ) : ℤ := kq - kp

/-- Remark 3.5.1 (why the asymmetric condition is necessary, C1 case):
    For p ∈ C1, if kq = kp, then q = p - 2 < p, and the product p*q has
    already been sieved by q. The strict inequality kq ≥ kp + 1 excludes
    this redundancy. Encoded as a purely arithmetic implication. -/
theorem why_asymmetric_c1 (p kp q kq : ℤ) (hp : p = 6 * kp + 1)
    (hq : q = 6 * kq - 1) (hkq : kq = kp) :
    q = p - 2 := by
  subst hkq; omega

/-- Thm 3.6 (N1-P): existence of Cop(p).
    The hypothesis `h_dirichlet` is the parameterised form of Dirichlet's
    theorem: there exists a prime q in the opposite channel whose index
    satisfies the asymmetric condition. This sanitises the analytic
    input (Dirichlet) as an explicit theorem hypothesis, eliminating
    any global axiom. -/
theorem cop_exists (p kp ep : ℤ)
    (h_dirichlet : ∃ q kq : ℤ, IsCop p kp ep q kq) :
    ∃ q kq : ℤ, IsCop p kp ep q kq := h_dirichlet

end ATM.Cop
""",

# ============================================================
# MODULE 6: DigitalFilter5 (N1)
# Thm 3.9 — Digital filter modulo 10
# ============================================================
"DigitalFilter5": """import Mathlib.Data.Int.Basic
import Mathlib.Tactic

namespace ATM.Filter5

/-- Thm 3.9 (mod 5, C1 channel): 5 | (6k + 1) - (k + 1).
    Equivalently, 6k + 1 ≡ k + 1 (mod 5). -/
theorem filter_mod_5_positive (k : ℤ) : 5 ∣ (6 * k + 1) - (k + 1) := by omega

/-- Thm 3.9 (mod 5, C5 channel): 5 | (6k - 1) - (k - 1).
    Equivalently, 6k - 1 ≡ k - 1 (mod 5). -/
theorem filter_mod_5_negative (k : ℤ) : 5 ∣ (6 * k - 1) - (k - 1) := by omega

/-- Corollary: the C1 reduction 5 | 6k + 1 ↔ 5 | k + 1. -/
theorem filter_c1_iff (k : ℤ) : (5 ∣ (6 * k + 1)) ↔ (5 ∣ (k + 1)) := by
  constructor
  · intro h
    have := filter_mod_5_positive k
    have hsub : k + 1 = (6 * k + 1) - ((6 * k + 1) - (k + 1)) := by ring
    rw [hsub]
    exact dvd_sub h this
  · intro h
    have := filter_mod_5_positive k
    have hadd : 6 * k + 1 = (k + 1) + ((6 * k + 1) - (k + 1)) := by ring
    rw [hadd]
    exact dvd_add h this

/-- Corollary: the C5 reduction 5 | 6k - 1 ↔ 5 | k - 1. -/
theorem filter_c5_iff (k : ℤ) : (5 ∣ (6 * k - 1)) ↔ (5 ∣ (k - 1)) := by
  constructor
  · intro h
    have := filter_mod_5_negative k
    have hsub : k - 1 = (6 * k - 1) - ((6 * k - 1) - (k - 1)) := by ring
    rw [hsub]
    exact dvd_sub h this
  · intro h
    have := filter_mod_5_negative k
    have hadd : 6 * k - 1 = (k - 1) + ((6 * k - 1) - (k - 1)) := by ring
    rw [hadd]
    exact dvd_add h this

end ATM.Filter5
""",

# ============================================================
# MODULE 7: TwinPrimeGroundState (N1-P)
# Thm 4.2 — Twin primes as ground state
# ============================================================
"TwinPrimeGroundState": """import Mathlib.Data.Nat.Basic
import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Int.Basic
import Mathlib.Tactic

namespace ATM.GroundState

/-- Twin-generation predicate: k generates a twin-prime pair iff both
    6k-1 and 6k+1 are prime. -/
def GeneraGemelos (k : ℕ) : Prop :=
  Nat.Prime (6 * k - 1) ∧ Nat.Prime (6 * k + 1)

/-- The quadratic annihilator (paper: eq:twin-annihilator). -/
def annihilator (k kp : ℤ) : ℤ := k ^ 2 - kp ^ 2

/-- Necessity: if k generates twins, then no prime p < 6k-1 divides
    either channel. -/
theorem condicion_necesaria_gemelos (k : ℕ) (hk : GeneraGemelos k) :
    ∀ p : ℕ, Nat.Prime p → p < 6 * k - 1 →
      ¬(p ∣ 6 * k - 1) ∧ ¬(p ∣ 6 * k + 1) := by
  intro p hp h_lt
  rcases hk with ⟨h_minus, h_plus⟩
  constructor
  · intro h_div
    have h_eq_or := (Nat.Prime.eq_one_or_self_of_dvd h_minus p) h_div
    rcases h_eq_or with rfl | rfl
    · exact Nat.not_prime_one hp
    · exact lt_irrefl _ h_lt
  · intro h_div
    have h_lt_plus : p < 6 * k + 1 := by omega
    have h_eq_or := (Nat.Prime.eq_one_or_self_of_dvd h_plus p) h_div
    rcases h_eq_or with rfl | rfl
    · exact Nat.not_prime_one hp
    · exact lt_irrefl _ h_lt_plus

/-- Sufficiency: if no prime p with p² ≤ 6k+1 divides either channel,
    then k generates twins. -/
theorem condicion_suficiente_gemelos (k : ℕ) (hk0 : 0 < k)
    (h_oraculo : ∀ p : ℕ, Nat.Prime p → p * p ≤ 6 * k + 1 →
      ¬(p ∣ 6 * k - 1) ∧ ¬(p ∣ 6 * k + 1)) :
    GeneraGemelos k := by
  have h_gt1_m : 1 < 6 * k - 1 := by omega
  have h_pos_m : 0 < 6 * k - 1 := by omega
  have h_gt1_p : 1 < 6 * k + 1 := by omega
  have h_pos_p : 0 < 6 * k + 1 := by omega
  constructor
  · by_contra h_comp
    have hp_prime := Nat.minFac_prime h_gt1_m.ne'
    have hp_dvd := Nat.minFac_dvd (6 * k - 1)
    have hp_sq : (Nat.minFac (6 * k - 1)) ^ 2 ≤ 6 * k - 1 :=
      Nat.minFac_sq_le_self h_pos_m h_comp
    have hp_le : (Nat.minFac (6 * k - 1)) * (Nat.minFac (6 * k - 1)) ≤ 6 * k + 1 := by
      calc (Nat.minFac (6 * k - 1)) * (Nat.minFac (6 * k - 1))
        _ = (Nat.minFac (6 * k - 1)) ^ 2 := by ring
        _ ≤ 6 * k - 1 := hp_sq
        _ ≤ 6 * k + 1 := by omega
    exact (h_oraculo (Nat.minFac (6 * k - 1)) hp_prime hp_le).left hp_dvd
  · by_contra h_comp
    have hp_prime := Nat.minFac_prime h_gt1_p.ne'
    have hp_dvd := Nat.minFac_dvd (6 * k + 1)
    have hp_sq : (Nat.minFac (6 * k + 1)) ^ 2 ≤ 6 * k + 1 :=
      Nat.minFac_sq_le_self h_pos_p h_comp
    have hp_le : (Nat.minFac (6 * k + 1)) * (Nat.minFac (6 * k + 1)) ≤ 6 * k + 1 := by
      calc (Nat.minFac (6 * k + 1)) * (Nat.minFac (6 * k + 1))
        _ = (Nat.minFac (6 * k + 1)) ^ 2 := by ring
        _ ≤ 6 * k + 1 := hp_sq
    exact (h_oraculo (Nat.minFac (6 * k + 1)) hp_prime hp_le).right hp_dvd

/-- Theorem thm:ground-state (N1): complete equivalence. -/
theorem equivalencia_gemelos (k : ℕ) (hk0 : 0 < k) :
    GeneraGemelos k ↔
    (∀ p : ℕ, Nat.Prime p → p * p ≤ 6 * k + 1 →
      ¬(p ∣ 6 * k - 1) ∧ ¬(p ∣ 6 * k + 1)) := by
  constructor
  · intro hk p hp h_sq
    rcases hk with ⟨hm, hp_prime⟩
    constructor
    · intro hdiv
      have h_eq := (Nat.Prime.eq_one_or_self_of_dvd hm p) hdiv
      rcases h_eq with rfl | rfl
      · exact Nat.not_prime_one hp
      · have hk_ge : 1 ≤ k := hk0
        have h_bound : 5 ≤ 6 * k - 1 := by omega
        have h_mul := Nat.mul_le_mul_left (6 * k - 1) h_bound
        omega
    · intro hdiv
      have h_eq := (Nat.Prime.eq_one_or_self_of_dvd hp_prime p) hdiv
      rcases h_eq with rfl | rfl
      · exact Nat.not_prime_one hp
      · have hk_ge : 1 ≤ k := hk0
        have h_bound : 7 ≤ 6 * k + 1 := by omega
        have h_mul := Nat.mul_le_mul_left (6 * k + 1) h_bound
        omega
  · intro h
    exact condicion_suficiente_gemelos k hk0 h

end ATM.GroundState
""",

# ============================================================
# MODULE 8: HardyLittlewood (N1-P)
# Thm 4.4 — HL local factor as survival density
# ============================================================
"HardyLittlewood": """import Mathlib.Basic.Real.Basic
import Mathlib.Tactic

namespace ATM.HL

noncomputable def localFactor (p : ℝ) : ℝ := 1 - 1 / (p - 1) ^ 2

/-- Thm 4.4 (algebraic identity): localFactor(p) = (p-2)*p / (p-1)^2 -/
theorem local_factor_identity (p : ℝ) (hp : p ≠ 1) :
    localFactor p = (p - 2) * p / (p - 1) ^ 2 := by
  unfold localFactor
  have h : (p - 1) ^ 2 ≠ 0 := by
    apply pow_ne_zero
    exact sub_ne_zero.mpr hp
  field_simp
  ring

/-- Thm 4.4 at p = 5 (numerical anchor) -/
theorem local_factor_at_5 :
    localFactor 5 = (5 - 2) * 5 / (5 - 1) ^ 2 := by
  apply local_factor_identity
  norm_num

end ATM.HL
""",

# ============================================================
# MODULE 9: Complexity (N1-P)
# Thm 5.1-5.2 — Asymptotic bounds
# ============================================================
"Complexity": """import Mathlib.Analysis.SpecialFunctions.Log.Basic
import Mathlib.Analysis.SpecialFunctions.Pow.Real
import Mathlib.Tactic

namespace ATM.Complexity

noncomputable def memoryBound (N : ℝ) : ℝ := Real.sqrt N / Real.log N
theorem memory_def (N : ℝ) : memoryBound N = Real.sqrt N / Real.log N := rfl

theorem memory_positive (N : ℝ) (hN : 1 < N) : 0 < memoryBound N := by
  unfold memoryBound
  apply div_pos
  · exact Real.sqrt_pos_of_pos (by linarith)
  · exact Real.log_pos hN

noncomputable def timeBound (N : ℝ) : ℝ := N ^ (3 / 2 : ℝ) / Real.log N
theorem time_def (N : ℝ) : timeBound N = N ^ (3 / 2 : ℝ) / Real.log N := rfl

end ATM.Complexity
""",

# ============================================================
# MODULE 10: SieveOperatorSelfAdjoint (N1)
# Remark — H_N = MM^T is symmetric
# ============================================================
"SieveOperatorSelfAdjoint": """import Mathlib.Data.Matrix.Basic
import Mathlib.Tactic

open Matrix

namespace ATM.SieveOperator

variable {k_idx p_idx : Type*} [Fintype k_idx] [Fintype p_idx]
variable {α : Type*} [CommRing α]

/-- The sieve operator H_N = M M^T is symmetric. -/
theorem sieve_operator_is_symmetric (M : Matrix k_idx p_idx α) :
    (M * M.transpose).transpose = M * M.transpose := by
  simp

end ATM.SieveOperator
""",

# ============================================================
# MODULE 11: OptimalityM6 (N1)
# Thm 6.1-6.2 — Structural optimality of m = 6
# ============================================================
"OptimalityM6": """import Mathlib.Data.Nat.Basic
import Mathlib.Data.Nat.Totient
import Mathlib.Data.Rat.Defs
import Mathlib.Tactic

namespace ATM.Optimality

def rho (m : ℕ) : ℚ := (Nat.totient m : ℚ) / (m : ℚ)

theorem totient_1 : Nat.totient 1 = 1 := by decide
theorem totient_2 : Nat.totient 2 = 1 := by decide
theorem totient_3 : Nat.totient 3 = 2 := by decide
theorem totient_4 : Nat.totient 4 = 2 := by decide
theorem totient_6 : Nat.totient 6 = 2 := by decide

theorem rho_1 : rho 1 = 1 := by unfold rho; rw [totient_1]; norm_num
theorem rho_2 : rho 2 = 1/2 := by unfold rho; rw [totient_2]; norm_num
theorem rho_3 : rho 3 = 2/3 := by unfold rho; rw [totient_3]; norm_num
theorem rho_4 : rho 4 = 1/2 := by unfold rho; rw [totient_4]; norm_num
theorem rho_6 : rho 6 = 1/3 := by unfold rho; rw [totient_6]; norm_num

/-- Thm 6.1: m = 6 uniquely minimises ρ(m) among φ(m) ≤ 2 -/
theorem rho_six_is_min :
    rho 6 ≤ rho 1 ∧ rho 6 ≤ rho 2 ∧ rho 6 ≤ rho 3 ∧ rho 6 ≤ rho 4 := by
  refine ⟨?_, ?_, ?_, ?_⟩ <;>
    simp [rho_1, rho_2, rho_3, rho_4, rho_6] <;> norm_num

/-- Thm 6.2: Euler factor identities (rational local factors) -/
theorem euler_factor_2 : (1 - 1/(2:ℚ)^2) = 3/4 := by norm_num
theorem euler_factor_3 : (1 - 1/(3:ℚ)^2) = 8/9 := by norm_num
theorem euler_product_6 : (3/4 : ℚ) * (8/9) = 2/3 := by norm_num

end ATM.Optimality
""",

# ============================================================
# MODULE 12: ChiralConvergence (N1-P)
# §7.3 — Chiral convergence c1/c5 → 1
# ============================================================
"ChiralConvergence": """import Mathlib.Basic.Real.Basic
import Mathlib.Tactic

namespace ATM.ChiralConvergence

/-- §7.3 (N1): algebraic core of the chiral convergence, c1/c5 = 1 + (c1-c5)/c5. -/
theorem chiral_ratio_identity
    (c1 c5 : ℕ → ℝ) (N : ℕ) (h : c5 N ≠ 0) :
    c1 N / c5 N = 1 + (c1 N - c5 N) / c5 N := by
  field_simp
  ring

/-- §7.3 (N1): reformulation as difference from 1. -/
theorem chiral_ratio_diff
    (c1 c5 : ℕ → ℝ) (N : ℕ) (h : c5 N ≠ 0) :
    c1 N / c5 N - 1 = (c1 N - c5 N) / c5 N := by
  rw [chiral_ratio_identity c1 c5 N h]
  ring

end ATM.ChiralConvergence
""",

"ThresholdSeparation": """import Mathlib.Data.Int.Basic
import Mathlib.Tactic

namespace ATM.ThresholdSeparation

/-- Thm 3.8 (i): p^2 < p*q for all 0 < p < q. -/
theorem kmin_plus_lt_pq (p q : ℤ) (hp : 0 < p) (hq : p < q) :
    p^2 < p * q := by
  have h1 : p^2 = p * p := by ring
  rw [h1]
  exact mul_lt_mul_of_pos_left hq hp

/-- Thm 3.8 (ii): for p ∈ C5 (p = 6 kp - 1), under kq ≥ kp,
    the asymmetric threshold strictly dominates the symmetric one. -/
theorem kmin_minus_gt_kmin_plus_C5
    (p kp kq : ℤ) (hp : p = 6 * kp - 1) (hkp : 0 < kp) (hkq : kp ≤ kq) :
    p * kq + kp > p * kp - kp := by
  have hp_pos : 0 < p := by omega
  have hdiff : p * kq + kp - (p * kp - kp) = p * (kq - kp) + 2 * kp := by ring
  have hkq_diff : 0 ≤ kq - kp := by omega
  have hnonneg : 0 ≤ p * (kq - kp) := mul_nonneg (le_of_lt hp_pos) hkq_diff
  have hkp2 : 0 < 2 * kp := by nlinarith
  nlinarith

/-- Thm 3.8 (iii): for p ∈ C1 (p = 6 kp + 1), under kq > kp,
    the asymmetric threshold strictly dominates the symmetric one. -/
theorem kmin_minus_gt_kmin_plus_C1
    (p kp kq : ℤ) (hp : p = 6 * kp + 1) (hkp : 0 < kp) (hkq : kp < kq) :
    p * kq - kp > p * kp + kp := by
  have hp_pos : 0 < p := by omega
  have hdiff : p * kq - kp - (p * kp + kp) = p * (kq - kp) - 2 * kp := by ring
  have hkq_diff : 0 < kq - kp := by omega
  have hprod_pos : 0 < p * (kq - kp) := mul_pos hp_pos hkq_diff
  have hkp2_pos : 0 < 2 * kp := by nlinarith
  nlinarith

end ATM.ThresholdSeparation
""",

"UnitGroupStructure": """import Mathlib.Data.Int.Basic
import Mathlib.Data.Nat.GCD.Basic
import Mathlib.Data.Finset.Basic
import Mathlib.Tactic

namespace ATM.UnitGroup

/-- §2.1: 5 is the involutive unit of Z/6Z (5 * 5 ≡ 1). -/
theorem involution_mul : (5 * 5) % 6 = 1 % 6 := by norm_num

/-- §2.1: 1 + 5 ≡ 0 mod 6 (chiral antisymmetry). -/
theorem involution_add : (1 + 5) % 6 = 0 % 6 := by norm_num

/-- §2.1: the unit group (Z/6Z)× is exactly {1, 5}. -/
theorem unit_group_is :
    (Finset.filter (fun x : ℕ => x.Coprime 6) (Finset.range 6)) = {1, 5} := by decide

end ATM.UnitGroup
""",

"EulerFactorization": """import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Real.Basic
import Mathlib.Tactic

namespace ATM.EulerFactorization

open Classical

/-- §2.1 (N1): every prime p > 3 lies in C1 or C5. -/
theorem domain_primes_mod_6 (p : ℕ) (hp : Nat.Prime p) (hp3 : 3 < p) :
    p % 6 = 1 ∨ p % 6 = 5 := by
  have h_mod : p % 6 = 0 ∨ p % 6 = 1 ∨ p % 6 = 2 ∨ p % 6 = 3 ∨ p % 6 = 4 ∨ p % 6 = 5 := by omega
  rcases h_mod with h0 | h1 | h2 | h3 | h4 | h5
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6), by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exact Or.inl h1
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6) + 1, by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exfalso
    have h_div : 3 ∣ p := ⟨2 * (p / 6) + 1, by omega⟩
    have hp_eq_3 : 3 = p := (hp.eq_one_or_self_of_dvd 3 h_div).resolve_left (by decide)
    omega
  · exfalso
    have h_div : 2 ∣ p := ⟨3 * (p / 6) + 2, by omega⟩
    have hp_eq_2 : 2 = p := (hp.eq_one_or_self_of_dvd 2 h_div).resolve_left (by decide)
    omega
  · exact Or.inr h5

/-- Euler factor at prime p. -/
noncomputable def eulerFactor (p : ℕ) (s : ℝ) : ℝ := (1 - (p : ℝ)^(-s))⁻¹

/-- Product over all primes up to N. -/
noncomputable def totalProd (s : ℝ) : ℕ → ℝ
| 0 => 1
| n + 1 => if Nat.Prime (n + 1) then totalProd s n * eulerFactor (n + 1) s else totalProd s n

/-- Product over primes in channels C1 ∪ C5 up to N. -/
noncomputable def prodChannels (s : ℝ) : ℕ → ℝ
| 0 => 1
| n + 1 => if Nat.Prime (n + 1) ∧ ((n + 1) % 6 = 1 ∨ (n + 1) % 6 = 5) then
             prodChannels s n * eulerFactor (n + 1) s
           else
             prodChannels s n

/-- §2.1 (N1): explicit inductive separation at k = 0. -/
theorem prod_eq_shift (k : ℕ) (s : ℝ) :
    totalProd s (k + 3) = prodChannels s (k + 3) * eulerFactor 2 s * eulerFactor 3 s := by
  induction k with
  | zero =>
    have ht0 : totalProd s 0 = 1 := rfl
    have ht1 : totalProd s 1 = 1 := by
      have eq : totalProd s 1 = if Nat.Prime 1 then totalProd s 0 * eulerFactor 1 s else totalProd s 0 := rfl
      rw [eq, if_neg Nat.not_prime_one, ht0]
    have ht2 : totalProd s 2 = eulerFactor 2 s := by
      have eq : totalProd s 2 = if Nat.Prime 2 then totalProd s 1 * eulerFactor 2 s else totalProd s 1 := rfl
      rw [eq, if_pos Nat.prime_two, ht1, one_mul]
    have ht3 : totalProd s 3 = eulerFactor 2 s * eulerFactor 3 s := by
      have eq : totalProd s 3 = if Nat.Prime 3 then totalProd s 2 * eulerFactor 3 s else totalProd s 2 := rfl
      rw [eq, if_pos Nat.prime_three, ht2]
    have hc0 : prodChannels s 0 = 1 := rfl
    have hc1 : prodChannels s 1 = 1 := by
      have eq : prodChannels s 1 = if Nat.Prime 1 ∧ (1 % 6 = 1 ∨ 1 % 6 = 5) then prodChannels s 0 * eulerFactor 1 s else prodChannels s 0 := rfl
      have h_neg : ¬ (Nat.Prime 1 ∧ (1 % 6 = 1 ∨ 1 % 6 = 5)) := fun h => Nat.not_prime_one h.1
      rw [eq, if_neg h_neg, hc0]
    have hc2 : prodChannels s 2 = 1 := by
      have eq : prodChannels s 2 = if Nat.Prime 2 ∧ (2 % 6 = 1 ∨ 2 % 6 = 5) then prodChannels s 1 * eulerFactor 2 s else prodChannels s 1 := rfl
      have h_neg : ¬ (Nat.Prime 2 ∧ (2 % 6 = 1 ∨ 2 % 6 = 5)) := fun h => by have := h.2; omega
      rw [eq, if_neg h_neg, hc1]
    have hc3 : prodChannels s 3 = 1 := by
      have eq : prodChannels s 3 = if Nat.Prime 3 ∧ (3 % 6 = 1 ∨ 3 % 6 = 5) then prodChannels s 2 * eulerFactor 3 s else prodChannels s 2 := rfl
      have h_neg : ¬ (Nat.Prime 3 ∧ (3 % 6 = 1 ∨ 3 % 6 = 5)) := fun h => by have := h.2; omega
      rw [eq, if_neg h_neg, hc2]
    rw [ht3, hc3]
    ring
  | succ n ih =>
    have eq1 : totalProd s (n + 4) = if Nat.Prime (n + 4) then totalProd s (n + 3) * eulerFactor (n + 4) s else totalProd s (n + 3) := rfl
    have eq2 : prodChannels s (n + 4) = if Nat.Prime (n + 4) ∧ ((n + 4) % 6 = 1 ∨ (n + 4) % 6 = 5) then prodChannels s (n + 3) * eulerFactor (n + 4) s else prodChannels s (n + 3) := rfl
    rw [eq1, eq2]
    by_cases hp : Nat.Prime (n + 4)
    · have h_gt : n + 4 > 3 := by omega
      have h_mod := domain_primes_mod_6 (n + 4) hp h_gt
      have c2 : Nat.Prime (n + 4) ∧ ((n + 4) % 6 = 1 ∨ (n + 4) % 6 = 5) := ⟨hp, h_mod⟩
      rw [if_pos hp, if_pos c2, ih]
      ring
    · have c2 : ¬ (Nat.Prime (n + 4) ∧ ((n + 4) % 6 = 1 ∨ (n + 4) % 6 = 5)) := fun h => hp h.1
      rw [if_neg hp, if_neg c2, ih]

/-- §2.1 (N1): finite Euler factorization (paper: rem:zeta, finite form). -/
theorem finite_euler_factorization_mod_6 (k : ℕ) (s : ℝ)
    (h2 : 1 - (2:ℝ)^(-s) ≠ 0) (h3 : 1 - (3:ℝ)^(-s) ≠ 0) :
    prodChannels s (k + 3) = totalProd s (k + 3) *
      (1 - (2:ℝ)^(-s)) * (1 - (3:ℝ)^(-s)) := by
  rw [prod_eq_shift k s]
  unfold eulerFactor
  generalize hA : (1 - (2:ℝ)^(-s)) = A
  generalize hB : (1 - (3:ℝ)^(-s)) = B
  have hA_neq : A ≠ 0 := by rwa [←hA]
  have hB_neq : B ≠ 0 := by rwa [←hB]
  calc
    prodChannels s (k + 3) = prodChannels s (k + 3) * 1 * 1 := by ring
    _ = prodChannels s (k + 3) * (A⁻¹ * A) * (B⁻¹ * B) := by
        rw [inv_mul_cancel₀ hA_neq, inv_mul_cancel₀ hB_neq]
    _ = prodChannels s (k + 3) * A⁻¹ * B⁻¹ * A * B := by ring

end ATM.EulerFactorization
""",

"CompositeStructure": """import Mathlib.Data.Int.Basic
import Mathlib.Data.Int.Basic
import Mathlib.RingTheory.Coprime.Basic
import Mathlib.Tactic.Ring
import Mathlib.Tactic.Linarith

namespace ATM.Composite

/-- Lema de Euclides vía Bézout: si p divide a 6*X y mcd(6, p) = 1, entonces p divide a X.
    Demostrado puramente mediante álgebra de anillos en ℤ con cero axiomas. -/
lemma dvd_of_dvd_mul_six (p X : ℤ)
    (h1 : p ∣ (6 * X))
    (hcoprime : IsCoprime (6 : ℤ) p) :
    p ∣ X := by
  unfold IsCoprime at hcoprime
  rcases hcoprime with ⟨a, b, hab⟩
  rcases h1 with ⟨c, hc⟩
  use a * c + b * X
  calc X
    _ = (a * 6 + b * p) * X := by rw [hab, one_mul]
    _ = a * (6 * X) + b * p * X := by ring
    _ = a * (p * c) + b * p * X := by rw [hc]
    _ = p * (a * c + b * X) := by ring

/-- Lema Fundamental de Congruencia Posicional:
    p divide al candidato 6k + eps si y solo si p divide a (k - ep*eps*kp). -/
lemma positional_congruence (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (hcop : IsCoprime (6 : ℤ) p) :
    p ∣ (6 * k + eps) ↔ p ∣ (k - ep * eps * kp) := by
  have h_alg : 6 * (k - ep * eps * kp) = (6 * k + eps) - p * (ep * eps) := by
    calc 6 * (k - ep * eps * kp)
      _ = 6 * k + eps - 6 * kp * ep * eps - 1 * eps := by ring
      _ = 6 * k + eps - 6 * kp * ep * eps - (ep * ep) * eps := by rw [hep]
      _ = 6 * k + eps - (6 * kp + ep) * (ep * eps) := by ring
      _ = (6 * k + eps) - p * (ep * eps) := by rw [hp]
  constructor
  · intro h_div
    rcases h_div with ⟨c, hc⟩
    have h_mul : p ∣ 6 * (k - ep * eps * kp) := by
      use c - ep * eps
      calc 6 * (k - ep * eps * kp)
        _ = (6 * k + eps) - p * (ep * eps) := h_alg
        _ = p * c - p * (ep * eps) := by rw [hc]
        _ = p * (c - ep * eps) := by ring
    exact dvd_of_dvd_mul_six p (k - ep * eps * kp) h_mul hcop
  · intro h_div
    rcases h_div with ⟨d, hd⟩
    use 6 * d + ep * eps
    calc 6 * k + eps
      _ = 6 * (k - ep * eps * kp) + p * (ep * eps) := by
        rw [h_alg]
        ring
      _ = 6 * (p * d) + p * (ep * eps) := by rw [hd]
      _ = p * (6 * d + ep * eps) := by ring

/-- Proposición 3.2 (Implicación Directa):
    Si p ∈ 𝒫_{>3} divide al candidato 6k+eps y es divisor propio (6k+eps ≠ p),
    entonces k pertenece necesariamente al conjunto de ramificación S_p. -/
theorem composite_forward_local
    (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (heps : eps = 1 ∨ eps = -1)
    (hep_cases : ep = 1 ∨ ep = -1)
    (hk : 0 < k)
    (_hkp : 0 < kp)
    (hp_pos : 0 < p)
    (hcop : IsCoprime (6 : ℤ) p)
    (h_div : p ∣ (6 * k + eps))
    (h_ne : 6 * k + eps ≠ p) :
    (eps = ep ∧ p ∣ (k - kp) ∧ k > kp) ∨ (eps = -ep ∧ p ∣ (k + kp)) := by
  have h_cong := (positional_congruence k kp p ep eps hp hep hcop).mp h_div
  by_cases h_sign : eps = ep
  · left
    have h_prod : ep * eps = 1 := by
      calc ep * eps
        _ = ep * ep := by rw [h_sign]
        _ = 1 := hep
    have h_dvd_sub : p ∣ (k - kp) := by
      have h_sub : k - ep * eps * kp = k - kp := by rw [h_prod, one_mul]
      rw [h_sub] at h_cong
      exact h_cong
    refine ⟨h_sign, h_dvd_sub, ?_⟩
    rcases h_dvd_sub with ⟨m, hm⟩
    by_contra h_not_gt
    have h_k_le : k ≤ kp := by omega
    have h_sub_le : k - kp ≤ 0 := by omega
    have h_m_le : m ≤ 0 := by
      by_contra hm_pos
      have hm_ge : 1 ≤ m := by omega
      have h_pm : p ≤ p * m := by nlinarith
      have : p ≤ 0 := by linarith [hm, h_pm, h_sub_le]
      linarith [hp_pos]
    have h_m_ge : 0 ≤ m := by
      by_contra hm_neg
      have hm_le : m ≤ -1 := by omega
      have h_pm : p * m ≤ -p := by nlinarith
      have h_k_le_neg : k - kp ≤ -p := by linarith [hm, h_pm]
      rcases hep_cases with h1 | h2
      · rw [h1] at hp; omega
      · rw [h2] at hp; omega
    have hm_zero : m = 0 := by omega
    have hk_eq : k = kp := by
      calc k
        _ = kp + (k - kp) := by ring
        _ = kp + p * m := by rw [hm]
        _ = kp := by rw [hm_zero, mul_zero, add_zero]
    have h_cand_eq : 6 * k + eps = p := by
      calc 6 * k + eps
        _ = 6 * kp + ep := by rw [hk_eq, h_sign]
        _ = p := hp.symm
    exact h_ne h_cand_eq
  · right
    have h_eps_neg : eps = -ep := by
      rcases heps with rfl | rfl <;> rcases hep_cases with rfl | rfl <;>
        (try rfl) <;> (exfalso; exact h_sign rfl)
    have h_prod_neg : ep * eps = -1 := by
      calc ep * eps
        _ = ep * (-ep) := by rw [h_eps_neg]
        _ = -(ep * ep) := by ring
        _ = -1 := by rw [hep]
    have h_dvd_add : p ∣ (k + kp) := by
      have h_add : k - ep * eps * kp = k + kp := by
        calc k - ep * eps * kp
          _ = k - (-1) * kp := by rw [h_prod_neg]
          _ = k + kp := by ring
      rw [h_add] at h_cong
      exact h_cong
    exact ⟨h_eps_neg, h_dvd_add⟩

/-- Proposición 3.2 (Implicación Recíproca):
    Si k pertenece al conjunto de ramificación S_p de un primo p ∈ 𝒫_{>3},
    entonces el candidato 6k+eps es un múltiplo propio no trivial de p. -/
theorem composite_backward_local
    (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (hk : 0 < k)
    (hkp : 0 < kp)
    (hep_cases : ep = 1 ∨ ep = -1)
    (h_branch : (eps = ep ∧ p ∣ (k - kp) ∧ k > kp) ∨ (eps = -ep ∧ p ∣ (k + kp)))
    (hcop : IsCoprime (6 : ℤ) p) :
    p ∣ (6 * k + eps) ∧ p < 6 * k + eps := by
  have hp_pos : 0 < p := by
    rcases hep_cases with h1 | h2
    · rw [h1] at hp; omega
    · rw [h2] at hp; omega
  rcases h_branch with ⟨h_same, h_dvd, h_gt⟩ | ⟨h_opp, h_dvd⟩
  · have h_prod : ep * eps = 1 := by
      calc ep * eps = ep * ep := by rw [h_same]
      _ = 1 := hep
    have h_dvd' : p ∣ (k - ep * eps * kp) := by
      have : k - ep * eps * kp = k - kp := by rw [h_prod, one_mul]
      rw [this]
      exact h_dvd
    have h_div_cand := (positional_congruence k kp p ep eps hp hep hcop).mpr h_dvd'
    have h_gt_cand : p < 6 * k + eps := by
      have hk_ge : kp + 1 ≤ k := by omega
      have h_eps_eq : eps = ep := h_same
      omega
    exact ⟨h_div_cand, h_gt_cand⟩
  · have h_prod : ep * eps = -1 := by
      calc ep * eps = ep * (-ep) := by rw [h_opp]
      _ = -(ep * ep) := by ring
      _ = -1 := by rw [hep]
    have h_dvd' : p ∣ (k - ep * eps * kp) := by
      have : k - ep * eps * kp = k + kp := by
        rw [h_prod]
        ring
      rw [this]
      exact h_dvd
    have h_div_cand := (positional_congruence k kp p ep eps hp hep hcop).mpr h_dvd'
    have h_gt_cand : p < 6 * k + eps := by
      rcases h_dvd with ⟨m, hm⟩
      have hm_pos : 1 ≤ m := by
        by_contra h_neg
        have hm_le : m ≤ 0 := by omega
        have h_pm : p * m ≤ 0 := by nlinarith [hp_pos, hm_le]
        have h_sum : 0 < k + kp := by omega
        linarith [hm, h_sum, h_pm]
      have h_pm_ge : p ≤ p * m := by nlinarith [hp_pos, hm_pos]
      have hk_ge : p - kp ≤ k := by linarith [hm, h_pm_ge]
      linarith [hk_ge, hp, h_opp, hp_pos]
    exact ⟨h_div_cand, h_gt_cand⟩

/-- Teorema Estructural 3.2 (Equivalencia Completa de Ramificación N1):
    Un índice k produce un candidato divisible por p y estrictamente compuesto
    si y solo si k pertenece a la rama correspondiente de S_p.
    Completamente verificado con 0 sorry y 0 axiomas. -/
theorem composite_branching_iff
    (k kp p ep eps : ℤ)
    (hp : p = 6 * kp + ep)
    (hep : ep * ep = 1)
    (heps : eps = 1 ∨ eps = -1)
    (hep_cases : ep = 1 ∨ ep = -1)
    (hk : 0 < k)
    (hkp : 0 < kp)
    (hp_pos : 0 < p)
    (hcop : IsCoprime (6 : ℤ) p) :
    (p ∣ (6 * k + eps) ∧ 6 * k + eps ≠ p) ↔
    ((eps = ep ∧ p ∣ (k - kp) ∧ k > kp) ∨ (eps = -ep ∧ p ∣ (k + kp))) := by
  constructor
  · intro ⟨h_div, h_ne⟩
    exact composite_forward_local k kp p ep eps hp hep heps hep_cases hk hkp hp_pos hcop h_div h_ne
  · intro h_branch
    have h_prop := composite_backward_local k kp p ep eps hp hep hk hkp hep_cases h_branch hcop
    exact ⟨h_prop.left, ne_of_gt h_prop.right⟩

end ATM.Composite
""",

"AnalyticOptimality": """import Mathlib.Analysis.SpecialFunctions.Pow.Real
import Mathlib.Analysis.SpecialFunctions.Pow.Real
import Mathlib.Analysis.SpecialFunctions.Trigonometric.Basic
import Mathlib.Tactic.Ring
import Mathlib.Tactic.NormNum

namespace ATM.AnalyticOptimality

/-- Interfaz algebraica mínima para caracteres de Dirichlet en factores de Euler. -/
structure Character where
  chi : ℤ → ℝ
  is_principal_mod_6 : Prop
  is_chi_neg4 : Prop
  is_chi_12 : Prop

/-- Factor local en p = 2 para el carácter principal módulo 6: (1 - 2^{-s}). -/
noncomputable def localFactorTwo (s : ℝ) : ℝ := 1 - (2 : ℝ)^(-s)

/-- Factor local en p = 3 para el carácter principal módulo 6: (1 - 3^{-s}). -/
noncomputable def localFactorThree (s : ℝ) : ℝ := 1 - (3 : ℝ)^(-s)

/-- §6.2 (N1): En s = 2, el factor local en p = 2 es exactamente 3/4. -/
theorem local_factor_2_at_2 : localFactorTwo 2 = 3/4 := by
  dsimp [localFactorTwo]
  rw [Real.rpow_neg (by norm_num)]
  rw [Real.rpow_two]
  norm_num

/-- §6.2 (N1): En s = 2, el factor local en p = 3 es exactamente 8/9. -/
theorem local_factor_3_at_2 : localFactorThree 2 = 8/9 := by
  dsimp [localFactorThree]
  rw [Real.rpow_neg (by norm_num)]
  rw [Real.rpow_two]
  norm_num

/-- §6.2 (N1): Corrección local combinada (3/4) * (8/9) = 2/3. -/
theorem local_factor_product_at_2 : localFactorTwo 2 * localFactorThree 2 = 2/3 := by
  rw [local_factor_2_at_2, local_factor_3_at_2]
  norm_num

/-- §6.2 (N1-P): Identidad algebraica analítica:
    L(2, χ_0^(6)) = [L(1, χ_12)]² = π²/9.
    Las evaluaciones clásicas ζ(2) = π²/6 y L(1, χ_{-4}) = π/4 se parametrizan como
    hipótesis locales explícitas, garantizando cero axiomas globales en el verificador. -/
theorem opt_analytic_identity
    (zeta_2 : ℝ) (L_chi_neg4 : ℝ)
    (h_zeta_2 : zeta_2 = Real.pi^2 / 6)
    (h_L_chi_neg4 : L_chi_neg4 = Real.pi / 4) :
    let L_2_chi_0_6 := zeta_2 * localFactorTwo 2 * localFactorThree 2
    let L_1_chi_12 := L_chi_neg4 * (4/3)
    L_2_chi_0_6 = L_1_chi_12^2 ∧ L_2_chi_0_6 = Real.pi^2 / 9 := by
  intro L_2_chi_0_6 L_1_chi_12
  refine ⟨?_, ?_⟩
  · show zeta_2 * localFactorTwo 2 * localFactorThree 2 = (L_chi_neg4 * (4/3))^2
    rw [h_zeta_2, h_L_chi_neg4, local_factor_2_at_2, local_factor_3_at_2]
    ring
  · show zeta_2 * localFactorTwo 2 * localFactorThree 2 = Real.pi^2 / 9
    rw [h_zeta_2, local_factor_2_at_2, local_factor_3_at_2]
    ring

end ATM.AnalyticOptimality
""",

"TwinPrimesExt": """import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Finset.Basic
import Mathlib.Algebra.BigOperators.Group.Finset.Basic
import Mathlib.Algebra.Order.BigOperators.Group.Finset
import Mathlib.Data.Real.Basic
import Mathlib.Tactic

open Finset

namespace ATM.TwinPrimesExt

-- ============================================================================
-- 1. DEFINICIONES BASE (DISCRETAS Y ASINTÓTICAS)
-- ============================================================================

/-- Base de números primos impares p ≥ 5 hasta la cota z. -/
def prime_basis (z : ℕ) : Finset ℕ :=
  (Finset.range (z + 1)).filter (fun p => 5 ≤ p ∧ Nat.Prime p)

/-- Primorial modificado M(z) = P(z)/6 sobre los primos p ≥ 5. -/
def M (z : ℕ) : ℕ :=
  ∏ p ∈ prime_basis z, p

/-- Cardinalidad combinatoria exacta de clases residuales supervivientes en ℕ. -/
def surviving_cardinality (z : ℕ) : ℕ :=
  ∏ p ∈ prime_basis z, (p - 2)

/-- Densidad estocástica clásica de coprimalidad doble en ℝ. -/
noncomputable def coprime_density (z : ℕ) : ℝ :=
  ∏ p ∈ prime_basis z, (1 - 1 / (p : ℝ)) ^ 2

/-- Factor singular de Hardy-Littlewood en ℝ (corrección determinista del aniquilador). -/
noncomputable def hardy_littlewood_factor (z : ℕ) : ℝ :=
  ∏ p ∈ prime_basis z, (1 - 1 / ((p : ℝ) - 1) ^ 2)

/-- Densidad exacta de supervivencia del aniquilador cuadrático en ℝ. -/
noncomputable def survival_density (z : ℕ) : ℝ :=
  ∏ p ∈ prime_basis z, (((p : ℝ) - 2) / (p : ℝ))

-- ============================================================================
-- 2. TEOREMAS DISCRETOS (POSITIVIDAD EN ℕ)
-- ============================================================================

/-- Teorema N1: La cardinalidad de candidatos a gemelos es estrictamente positiva
    dentro del periodo fundamental CRT para todo z ≥ 5. -/
theorem surviving_cardinality_pos (z : ℕ) (hz : 5 ≤ z) : 0 < surviving_cardinality z := by
  dsimp [surviving_cardinality]
  apply prod_pos
  intro p hp
  rw [prime_basis, mem_filter, mem_range] at hp
  omega

/-- Teorema N1: El primorial modificado M(z) es estrictamente positivo. -/
theorem M_pos (z : ℕ) : 0 < M z := by
  dsimp [M]
  apply prod_pos
  intro p hp
  rw [prime_basis, mem_filter, mem_range] at hp
  omega

-- ============================================================================
-- 3. LEMAS AUXILIARES DE ACOTACIÓN EN ℝ
-- ============================================================================

lemma p_ge_five (p : ℕ) (hp : p ∈ prime_basis z) : (5 : ℝ) ≤ (p : ℝ) := by
  rw [prime_basis, mem_filter, mem_range] at hp
  exact Nat.cast_le.mpr hp.2.1

lemma p_gt_one (p : ℕ) (hp : p ∈ prime_basis z) : (1 : ℝ) < (p : ℝ) := by
  have := p_ge_five p hp
  linarith

lemma p_ne_zero (p : ℕ) (hp : p ∈ prime_basis z) : (p : ℝ) ≠ 0 := by
  have := p_ge_five p hp
  linarith

lemma p_ne_one (p : ℕ) (hp : p ∈ prime_basis z) : (p : ℝ) ≠ 1 := by
  have := p_ge_five p hp
  linarith

-- ============================================================================
-- 4. IDENTIDAD LOCAL Y FACTORIZACIÓN DE HARDY-LITTLEWOOD EN ℝ
-- ============================================================================

/-- Teorema N1: Identidad algebraica local de Hardy-Littlewood sobre un cuerpo.
    Demostrada puramente mediante normalización de cuerpos con cero axiomas. -/
theorem hl_local_identity (p : ℝ) (hp0 : p ≠ 0) (hp1 : p ≠ 1) :
    (p - 2) / p = (1 - 1 / p) ^ 2 * (1 - 1 / (p - 1) ^ 2) := by
  have h1 : p - 1 ≠ 0 := sub_ne_zero.mpr hp1
  have hsq1 : (1 - 1 / p) ^ 2 = (1 - 1 / p) * (1 - 1 / p) := by ring
  have hsq2 : (p - 1) ^ 2 = (p - 1) * (p - 1) := by ring
  rw [hsq1, hsq2]
  field_simp
  ring

/-- Teorema N1: Factorización global de la densidad de supervivencia.
    La densidad en el sub-retículo se factoriza idénticamente en la densidad
    estocástica por el producto de factores de Hardy-Littlewood. -/
theorem density_factorization (z : ℕ) :
    survival_density z = coprime_density z * hardy_littlewood_factor z := by
  dsimp [survival_density, coprime_density, hardy_littlewood_factor]
  have h_step : ∏ p ∈ prime_basis z, (((p : ℝ) - 2) / (p : ℝ)) =
                ∏ p ∈ prime_basis z, ((1 - 1 / (p : ℝ)) ^ 2 * (1 - 1 / ((p : ℝ) - 1) ^ 2)) := by
    apply prod_congr rfl
    intro p hp
    exact hl_local_identity (p : ℝ) (p_ne_zero p hp) (p_ne_one p hp)
  rw [h_step]
  exact prod_mul_distrib

/-- Teorema N1: Evaluación numérica puntual en el anclaje p = 5. -/
theorem hl_local_factor_at_5 :
    ((5 : ℝ) - 2) / 5 = (1 - 1 / (5 : ℝ)) ^ 2 * (1 - 1 / ((5 : ℝ) - 1) ^ 2) := by
  exact hl_local_identity 5 (by norm_num) (by norm_num)

-- ============================================================================
-- 5. RELACIÓN MULTIPLICATIVA ENTRE EL PRIMORIAL Y LOS SUPERVIVIENTES
-- ============================================================================

/-- Teorema N1: Relación canónica entre la densidad continua y las cantidades discretas.
    M(z) * survival_density(z) = ∏ (p - 2). -/
theorem survival_density_mul_M (z : ℕ) :
    (∏ p ∈ prime_basis z, (p : ℝ)) * survival_density z = ∏ p ∈ prime_basis z, ((p : ℝ) - 2) := by
  dsimp [survival_density]
  rw [← prod_mul_distrib]
  apply prod_congr rfl
  intro p hp
  exact mul_div_cancel₀ ((p : ℝ) - 2) (p_ne_zero p hp)

-- ============================================================================
-- 6. POSITIVIDAD DE LOS FACTORES ASINTÓTICOS
-- ============================================================================

lemma coprime_factor_pos (p : ℕ) (hp : p ∈ prime_basis z) :
    0 < (1 - 1 / (p : ℝ)) ^ 2 := by
  have hp0 : 0 < (p : ℝ) := by have := p_ge_five p hp; linarith
  have h_frac : 1 / (p : ℝ) < 1 := by
    rw [div_lt_iff₀ hp0]
    have := p_ge_five p hp
    linarith
  have h_sub : 0 < 1 - 1 / (p : ℝ) := by linarith
  have h_sq : (1 - 1 / (p : ℝ)) ^ 2 = (1 - 1 / (p : ℝ)) * (1 - 1 / (p : ℝ)) := by ring
  rw [h_sq]
  exact mul_pos h_sub h_sub

lemma hl_factor_pos (p : ℕ) (hp : p ∈ prime_basis z) :
    0 < 1 - 1 / ((p : ℝ) - 1) ^ 2 := by
  have hp5 := p_ge_five p hp
  have h_p1_pos : 0 < (p : ℝ) - 1 := by linarith
  have h_sq_pos : 0 < ((p : ℝ) - 1) ^ 2 := by
    have h_sq : ((p : ℝ) - 1) ^ 2 = ((p : ℝ) - 1) * ((p : ℝ) - 1) := by ring
    rw [h_sq]
    exact mul_pos h_p1_pos h_p1_pos
  have h_frac : 1 / ((p : ℝ) - 1) ^ 2 < 1 := by
    rw [div_lt_iff₀ h_sq_pos]
    nlinarith
  linarith

/-- Teorema N1: La densidad coprima doble es estrictamente positiva. -/
theorem coprime_density_pos (z : ℕ) : 0 < coprime_density z := by
  dsimp [coprime_density]
  apply prod_pos
  intro p hp
  exact coprime_factor_pos p hp

/-- Teorema N1: El factor compuesto de Hardy-Littlewood es estrictamente positivo. -/
theorem hardy_littlewood_factor_pos (z : ℕ) : 0 < hardy_littlewood_factor z := by
  dsimp [hardy_littlewood_factor]
  apply prod_pos
  intro p hp
  exact hl_factor_pos p hp

/-- Teorema N1: La densidad total de supervivencia al aniquilador cuadrático
    es estrictamente positiva en ℝ para cualquier escala z. -/
theorem survival_density_pos (z : ℕ) : 0 < survival_density z := by
  rw [density_factorization z]
  exact mul_pos (coprime_density_pos z) (hardy_littlewood_factor_pos z)

end ATM.TwinPrimesExt
""",

"TwinPrimesBoth": """import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Nat.Prime.Basic
import Mathlib.Data.Int.Basic
import Mathlib.Data.Finset.Basic
import Mathlib.Algebra.BigOperators.Group.Finset.Basic
import Mathlib.Data.Real.Basic
import Mathlib.Tactic

namespace ATM.Channel

/-- El canal C1 está bloqueado en el índice k por el primo p. -/
def B_C1 (k : ℤ) (p : ℤ) : Prop := p ∣ (6 * k + 1)

/-- El canal C5 está bloqueado en el índice k por el primo p. -/
def B_C5 (k : ℤ) (p : ℤ) : Prop := p ∣ (6 * k - 1)

/-- La involución quiral transforma un bloqueo en C1 en k en un bloqueo en C5 en -k. -/
theorem chiral_involution_C1_C5 (k : ℤ) (p : ℤ) :
    B_C1 k p ↔ B_C5 (-k) p := by
  dsimp [B_C1, B_C5]
  constructor
  · rintro ⟨c, hc⟩
    refine ⟨-c, ?_⟩
    calc 6 * (-k) - 1
      _ = -(6 * k + 1) := by ring
      _ = -(p * c) := by rw [hc]
      _ = p * (-c) := by ring
  · rintro ⟨c, hc⟩
    refine ⟨-c, ?_⟩
    calc 6 * k + 1
      _ = -(6 * (-k) - 1) := by ring
      _ = -(p * c) := by rw [hc]
      _ = p * (-c) := by ring

/-- La involución quiral transforma un bloqueo en C5 en k en un bloqueo en C1 en -k. -/
theorem chiral_involution_C5_C1 (k : ℤ) (p : ℤ) :
    B_C5 k p ↔ B_C1 (-k) p := by
  dsimp [B_C1, B_C5]
  constructor
  · rintro ⟨c, hc⟩
    refine ⟨-c, ?_⟩
    calc 6 * (-k) + 1
      _ = -(6 * k - 1) := by ring
      _ = -(p * c) := by rw [hc]
      _ = p * (-c) := by ring
  · rintro ⟨c, hc⟩
    refine ⟨-c, ?_⟩
    calc 6 * k - 1
      _ = -(6 * (-k) + 1) := by ring
      _ = -(p * c) := by rw [hc]
      _ = p * (-c) := by ring

end ATM.Channel

namespace ATM.TwinPrimesBoth

open Finset
open ATM.Channel (B_C1 B_C5 chiral_involution_C1_C5 chiral_involution_C5_C1)

-- ============================================================================
-- 1. BLOQUEO BILATERAL: DEFINICIÓN Y ESTRUCTURA
-- ============================================================================

/-- Bloqueo bilateral: ambos canales están bloqueados por primos de la base. -/
def B_both (k : ℤ) (primes : Finset ℕ) : Prop :=
  (∃ p ∈ primes, B_C1 k (p : ℤ)) ∧ (∃ q ∈ primes, B_C5 k (q : ℤ))

/-- Teorema N1: un único primo p ≥ 5 jamás bloquea ambos canales a la vez. -/
theorem single_prime_cannot_block_both (k : ℤ) (p : ℤ) (hp : 5 ≤ p) :
    ¬ (B_C1 k p ∧ B_C5 k p) := by
  dsimp [B_C1, B_C5]
  rintro ⟨h1, h2⟩
  have h_diff : p ∣ (6 * k + 1) - (6 * k - 1) := dvd_sub h1 h2
  have h_eq : (6 * k + 1) - (6 * k - 1) = 2 := by ring
  rw [h_eq] at h_diff
  rcases h_diff with ⟨c, hc⟩
  have h_c_pos : 0 < c := by
    by_contra! h_le
    have : p * c ≤ 0 := by nlinarith
    linarith
  have h_ge : 5 ≤ p * c := by nlinarith
  linarith

/-- Teorema N1: todo bloqueo bilateral exige dos primos distintos. -/
theorem bilateral_block_requires_distinct_primes
    (k : ℤ) (primes : Finset ℕ)
    (h_ge5 : ∀ p ∈ primes, 5 ≤ (p : ℤ))
    (p q : ℕ) (hp : p ∈ primes) (hq : q ∈ primes)
    (h_c1 : B_C1 k (p : ℤ)) (h_c5 : B_C5 k (q : ℤ)) :
    p ≠ q := by
  rintro rfl
  exact single_prime_cannot_block_both k (p : ℤ) (h_ge5 p hp) ⟨h_c1, h_c5⟩

/-- Teorema N1: para la base singleton {5}, el bloqueo bilateral es imposible. -/
theorem no_bilateral_block_singleton_five (k : ℤ) :
    ¬ B_both k {5} := by
  rintro ⟨⟨p, hp, h_c1⟩, ⟨q, hq, h_c5⟩⟩
  rw [mem_singleton] at hp hq
  subst hp hq
  exact single_prime_cannot_block_both k 5 (by norm_num) ⟨h_c1, h_c5⟩

-- ============================================================================
-- 2. JERARQUÍA LÓGICA DE INCLUSIÓN QUIRAL
-- ============================================================================

/-- Teorema N1: el bloqueo bilateral implica lógicamente el bloqueo en C1. -/
theorem both_implies_C1 (k : ℤ) (primes : Finset ℕ) :
    B_both k primes → (∃ p ∈ primes, B_C1 k (p : ℤ)) :=
  fun h => h.1

/-- Teorema N1: el bloqueo bilateral implica lógicamente el bloqueo en C5. -/
theorem both_implies_C5 (k : ℤ) (primes : Finset ℕ) :
    B_both k primes → (∃ q ∈ primes, B_C5 k (q : ℤ)) :=
  fun h => h.2

/-- Teorema N1: la involución quiral k ↦ -k permuta los testigos del bloqueo
    bilateral, intercambiando C1 y C5 sin ambigüedad de tipos. -/
theorem bilateral_block_chiral_involution (k : ℤ) (primes : Finset ℕ) :
    B_both k primes ↔ B_both (-k) primes := by
  dsimp [B_both]
  constructor
  · rintro ⟨⟨p, hp, h1⟩, ⟨q, hq, h2⟩⟩
    refine ⟨⟨q, hq, (chiral_involution_C5_C1 k (q : ℤ)).mp h2⟩,
            ⟨p, hp, (chiral_involution_C1_C5 k (p : ℤ)).mp h1⟩⟩
  · rintro ⟨⟨q, hq, h1⟩, ⟨p, hp, h2⟩⟩
    refine ⟨⟨p, hp, (chiral_involution_C1_C5 k (p : ℤ)).mpr h2⟩,
            ⟨q, hq, (chiral_involution_C5_C1 k (q : ℤ)).mpr h1⟩⟩

-- ============================================================================
-- 3. JERARQUÍA DE RACHAS: DE LO BILATERAL A LO MONO-CANAL
-- ============================================================================

/-- Una racha bilateral de longitud L a partir del índice k0. -/
def is_bilateral_run (primes : Finset ℕ) (k0 L : ℤ) : Prop :=
  0 < L ∧ ∀ j, 0 ≤ j → j < L → B_both (k0 + j) primes

/-- Una racha mono-canal C1 de longitud L a partir del índice k0. -/
def is_C1_run (primes : Finset ℕ) (k0 L : ℤ) : Prop :=
  0 < L ∧ ∀ j, 0 ≤ j → j < L → ∃ p ∈ primes, B_C1 (k0 + j) (p : ℤ)

/-- Una racha mono-canal C5 de longitud L a partir del índice k0. -/
def is_C5_run (primes : Finset ℕ) (k0 L : ℤ) : Prop :=
  0 < L ∧ ∀ j, 0 ≤ j → j < L → ∃ q ∈ primes, B_C5 (k0 + j) (q : ℤ)

/-- Teorema N1: toda racha bilateral de longitud L contiene una racha C1
    de la misma longitud. -/
theorem bilateral_run_implies_C1_run
    (primes : Finset ℕ) (k0 L : ℤ) (h : is_bilateral_run primes k0 L) :
    is_C1_run primes k0 L := by
  rcases h with ⟨hL, hrun⟩
  refine ⟨hL, ?_⟩
  intro j hj0 hjL
  exact both_implies_C1 (k0 + j) primes (hrun j hj0 hjL)

/-- Teorema N1: toda racha bilateral de longitud L contiene una racha C5
    de la misma longitud. -/
theorem bilateral_run_implies_C5_run
    (primes : Finset ℕ) (k0 L : ℤ) (h : is_bilateral_run primes k0 L) :
    is_C5_run primes k0 L := by
  rcases h with ⟨hL, hrun⟩
  refine ⟨hL, ?_⟩
  intro j hj0 hjL
  exact both_implies_C5 (k0 + j) primes (hrun j hj0 hjL)

-- ============================================================================
-- 4. VENTANA DE PRIMALIDAD Y DOMINANCIA ASINTÓTICA
-- ============================================================================

/-- Escala canónica del intervalo de búsqueda de primos: X ≈ z^2 / 6. -/
noncomputable def X_window (z : ℝ) : ℝ := (z ^ 2) / 6

/-- Teorema N1: la ventana es estrictamente positiva para z > 0. -/
theorem X_window_pos (z : ℝ) (hz : 0 < z) : 0 < X_window z := by
  dsimp [X_window]
  have : 0 < z ^ 2 := sq_pos_of_pos hz
  linarith

/-- Teorema N1: la ventana X(z) domina asintóticamente a cualquier cota
    lineal Cz. -/
theorem X_window_eventually_dominates_linear (C : ℝ) :
    ∃ z0 : ℝ, ∀ z ≥ z0, C * z < X_window z := by
  refine ⟨|6 * C| + 1, ?_⟩
  intro z hz
  have hz_pos : 0 < z := by
    have : 0 ≤ |6 * C| := abs_nonneg _
    linarith
  dsimp [X_window]
  have h_bound : 6 * C < z := by
    have h_le : 6 * C ≤ |6 * C| := le_abs_self _
    linarith
  have h_mul : 6 * C * z < z * z := mul_lt_mul_of_pos_right h_bound hz_pos
  have h_sq : z * z = z ^ 2 := by ring
  rw [h_sq] at h_mul
  linarith

end ATM.TwinPrimesBoth
""",
}

for name, code in MODULES.items():
    with open(os.path.join(SRC_DIR, f"{name}.lean"), "w", encoding="utf-8") as f:
        f.write(code)

print(f"✅ All {len(MODULES)} formalization modules have been created on disk.")

# ------------------------------------------------------------------------------
# 2. LAKEFILE CONFIGURATION
# ------------------------------------------------------------------------------
lakefile_path = os.path.join(PROJECT_DIR, "lakefile.toml")
if os.path.exists(lakefile_path):
    with open(lakefile_path, "r", encoding="utf-8") as f:
        toml_content = f.read()
    if "globs" not in toml_content:
        toml_content = toml_content.replace(
            '[[lean_lib]]\nname = "ATM_Lean"',
            '[[lean_lib]]\nname = "ATM_Lean"\nglobs = ["ATM_Lean.+"]'
        )
        with open(lakefile_path, "w", encoding="utf-8") as f:
            f.write(toml_content)

# Root file linking all modules
ROOT_FILE = os.path.join(PROJECT_DIR, "ATM_Lean.lean")
root_modules = sorted(list(MODULES.keys()))
content = "\n".join([f"import ATM_Lean.{mod}" for mod in root_modules]) + "\n"
with open(ROOT_FILE, "w", encoding="utf-8") as f:
    f.write(content)

# ------------------------------------------------------------------------------
# 3. CLEANING RESIDUAL LOCAL BINARIES
# ------------------------------------------------------------------------------
local_build_dir = os.path.join(PROJECT_DIR, ".lake", "build", "lib", "ATM_Lean")
if os.path.exists(local_build_dir):
    shutil.rmtree(local_build_dir)
    print("🧹 Residual local binaries deleted.")

# ------------------------------------------------------------------------------
# 4. EXHAUSTIVE COMPILATION AND STRICT AUDIT
# ------------------------------------------------------------------------------
print("\n" + "=" * 95)
print("INTEGRITY AUDIT: GLOBAL AND FILE-BY-FILE VERIFICATION")
print("=" * 95)
print(f"Timestamp : {datetime.now(timezone.utc).strftime('%Y-%m-%d %H:%M:%S UTC')}")
print("Compiler  : Lean 4 (x86_64-unknown-linux-gnu)")
print("=" * 95)

print("\n[Step 1/2] Compiling library graph with Lake...")
build_res = subprocess.run(
    f"PATH='{env_path}' lake build ATM_Lean",
    shell=True, cwd=PROJECT_DIR, capture_output=True, text=True
)

if build_res.returncode != 0:
    print(f"❌ Error in lake build:\n{build_res.stderr}", file=sys.stderr)
    print(f"\n--- stdout ---\n{build_res.stdout}", file=sys.stderr)
    raise RuntimeError("Global Lake compilation failed.")
else:
    print("✅ lake build completed successfully (zero compilation errors).\n")

print("[Step 2/2] Independent kernel verification file by file...")

def hash_file(filepath):
    hasher = hashlib.sha256()
    with open(filepath, 'rb') as f:
        hasher.update(f.read())
    return hasher.hexdigest()

def analyze_lean_file(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        code = f.read()
    no_block = re.sub(r'/-[\s\S]*?-/', '', code)
    clean = re.sub(r'--.*', '', no_block)
    has_sorry = bool(re.search(r'\bsorry\b(?!_placeholder)', clean))
    has_placeholder = bool(re.search(r'sorry_placeholder', clean))
    axioms = re.findall(r'\baxiom\s+([a-zA-Z0-9_]+)', clean)
    theorems = re.findall(r'\b(?:theorem|lemma|proposition)\s+([a-zA-Z0-9_]+)', clean)
    defs = re.findall(r'\b(?:noncomputable\s+)?def\s+([a-zA-Z0-9_]+)', clean)
    if has_sorry or has_placeholder:
        level = "N2 (sorry)"
    elif len(axioms) > 0:
        level = f"N2 ({len(axioms)} ax)"
    elif "N1-P" in code:
        level = "N1-P"
    else:
        level = "N1"
    return level, len(theorems), len(defs)

lean_files = sorted([f for f in os.listdir(SRC_DIR) if f.endswith('.lean') and not f.startswith('_')])
n1_count, n1p_count, n2_count = 0, 0, 0
total_thms, total_defs = 0, 0
failures = []

print(f"{'Module':<32} | {'Level':<11} | {'Thms':<4} | {'Defs':<4} | {'SHA-256 (first 32)':<34}")
print("-" * 105)

for name in lean_files:
    filepath = os.path.join(SRC_DIR, name)
    level, thms, defs = analyze_lean_file(filepath)
    file_hash = hash_file(filepath)

    res = subprocess.run(
        f"PATH='{env_path}' lake env lean {filepath}",
        shell=True, cwd=PROJECT_DIR, capture_output=True, text=True, timeout=120
    )

    if res.returncode == 0:
        status = "✅"
        if "N1-P" in level:
            n1p_count += 1
        elif "N1" in level:
            n1_count += 1
        else:
            n2_count += 1
        total_thms += thms
        total_defs += defs
    else:
        status = "❌"
        raw_err = res.stdout + res.stderr
        clean_err = re.sub(r'\x1b\[[0-9;]*m', '', raw_err).strip()
        failures.append((name, clean_err))

    print(f"{status} {name:<30} | {level:<11} | {thms:>3} | {defs:>3} | {file_hash[:32]}")

print("-" * 105)
if failures:
    print(f"❌ {len(failures)} modules failed in the kernel.", file=sys.stderr)
    for f_name, f_err in failures:
        print(f"\n=== Failure in {f_name} ===\n{f_err[:800]}", file=sys.stderr)
    raise RuntimeError("Audit failed.")
else:
    print(f"SUMMARY: {n1_count} Modules at N1 | {n1p_count} Modules at N1-P | {n2_count} Modules at N2")
    print(f"TOTAL: {total_thms} certified theorems and {total_defs} verified definitions.")

# ------------------------------------------------------------------------------
# 5. REPRODUCIBLE ZIP PACKAGE
# ------------------------------------------------------------------------------
config_files = ["lakefile.lean", "lakefile.toml", "lean-toolchain", "lake-manifest.json", "ATM_Lean.lean"]
with zipfile.ZipFile(ZIP_OUTPUT_PATH, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for cfg in config_files:
        cfg_path = os.path.join(PROJECT_DIR, cfg)
        if os.path.exists(cfg_path):
            zipf.write(cfg_path, arcname=os.path.join("ATM_Lean", cfg))
    for root, _, files in os.walk(SRC_DIR):
        for file in sorted(files):
            if file.endswith(".lean") and not file.startswith("_"):
                full_path = os.path.join(root, file)
                rel_path = os.path.relpath(full_path, PROJECT_DIR)
                zipf.write(full_path, arcname=os.path.join("ATM_Lean", rel_path))

with open(ZIP_OUTPUT_PATH, 'rb') as f:
    zip_hash = hashlib.sha256(f.read()).hexdigest()

print(f"\n✅ REPRODUCIBILITY SUITE: {ZIP_OUTPUT_PATH}")
print(f"   Size   : {os.path.getsize(ZIP_OUTPUT_PATH)/1024.0:.2f} KB")
print(f"   SHA-256: {zip_hash}")
```
